# Long-Term Memory Across Sessions [Step 1 - Persistent User Profiles]

> **MLCourse - Agentic AI - Persistent Memory RAG**

Chat models are stateless: every call starts from zero. To remember a user
across sessions, you need persistent storage. This notebook builds a system
that stores user preferences, past queries, and learned facts in a
`InMemoryStore`, then retrieves them in a brand-new session.

### What you will learn

1. Why per-session memory is not enough for real applications
2. How `InMemoryStore` provides cross-thread, cross-session persistence
3. Storing and retrieving user profiles by namespace
4. User isolation: Alice never sees Bob's preferences
5. Updating stored facts over time as the user shares new information

```python
# --- Section 2: the session-only memory problem --------------------------------
# Without persistence, a chatbot forgets everything when the session ends.
# We simulate this by clearing the message list between sessions.

def run_session_no_memory(session_name, turns):
    """Simulate a session with NO persistent memory."""
    history = []          # ephemeral list -- lost when function returns
    print("--- %s (no memory) ---" % session_name)
    for user_text in turns:
        history.append(user_text)
        reply = "[simulated reply to: %s]" % user_text
        print("  user: %s" % user_text)
        print("  bot : %s" % reply)
    print("  messages kept: %d (all ephemeral)" % len(history))
    return history

# Turn 1: user shares preferences
session1 = run_session_no_memory("Session 1", [
    "Hi, my name is Alice.",
    "My favorite color is purple.",
    "I enjoy hiking and painting.",
])

# Session 2: bot has NO memory of Session 1
session2 = run_session_no_memory("Session 2", [
    "Hello again! Do you remember me?",
])
print("\nlesson: the bot in Session 2 has zero knowledge of Alice's preferences.")
print("cross-session memory requires persistent storage.\n")
```

```python
# --- Section 3: introducing InMemoryStore --------------------------------------
# InMemoryStore holds key-value items organized by namespace (tuple of strings).
# It lives in RAM but persists across threads and sessions within the same process.
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()
print("InMemoryStore created, type:", type(store).__name__)

# Manual store/get demo
store.put(("user", "alice"), "profile", {"name": "Alice", "color": "purple"})
item = store.get(("user", "alice"), "profile")
print("Stored and retrieved Alice's profile:", item.value)

# Namespace isolation
store.put(("user", "bob"), "profile", {"name": "Bob", "color": "green"})
all_users = store.search(("user",))
print("\nAll stored user profiles:")
for u in all_users:
    print("  %s -> %s" % (u.key, u.value))
```

```python
# --- Section 4: building a memory-augmented chatbot node -----------------------
# This function reads user profile from the store, injects it as context,
# calls the LLM, then extracts and saves any new facts from the user's message.
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage
from typing import TypedDict, Annotated
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import MemorySaver

# Re-initialize a clean store for the graph
store = InMemoryStore()

class ChatState(TypedDict):
    messages: Annotated[list, add_messages]

memory_prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are a friendly assistant with reliable long-term memory. "
     "You remember the user's name, preferences, and past conversations. "
     "When the user shares a fact about themselves, acknowledge it.\n\n"
     "Known user info:\n{user_context}"),
    MessagesPlaceholder(variable_name="messages"),
])

def extract_facts(message_text):
    """Extract structured facts from a user message. Returns a dict of new facts."""
    facts = {}
    text_lower = message_text.lower()

    if "my name is" in text_lower:
        match = re.search(r"my name is\s+(\w+)", text_lower)
        if match:
            facts["name"] = match.group(1).title()

    if "favorite color" in text_lower:
        for kw in ["is", "my"]:
            if kw in text_lower:
                idx = text_lower.index(kw) + len(kw)
                rest = message_text[idx:].strip().split()
                if rest:
                    facts["favorite_color"] = rest[0].strip(".,!?").lower()
                break

    if "i enjoy" in text_lower or "i like" in text_lower:
        for kw in ["i enjoy", "i like"]:
            if kw in text_lower:
                idx = text_lower.index(kw) + len(kw)
                facts["interests"] = message_text[idx:].strip().rstrip(".")
                break

    return facts

def chatbot_with_memory(state: ChatState, config):
    """Chatbot node that reads/writes to persistent store."""
    user_ns = ("user", config["configurable"].get("user_id", "default"))

    # Read existing profile from store
    profile_item = store.get(user_ns, "profile")
    profile = profile_item.value if profile_item else {}

    # Inject profile as context
    context_lines = []
    if profile.get("name"):
        context_lines.append("Name: %s" % profile["name"])
    if profile.get("favorite_color"):
        context_lines.append("Favorite color: %s" % profile["favorite_color"])
    if profile.get("interests"):
        context_lines.append("Interests: %s" % profile["interests"])
    user_context = "\n".join(context_lines) if context_lines else "(no info yet)"

    # Build messages with context
    msgs = memory_prompt.format_messages(
        user_context=user_context,
        messages=state["messages"],
    )

    if LLM_LIVE:
        response = llm.invoke(msgs)
    else:
        # Offline stub: echo what we know about the user
        last_msg = state["messages"][-1].content if state["messages"] else ""
        if profile.get("name"):
            name = profile["name"]
            if "favorite color" in last_msg.lower():
                resp_text = "Your favorite color is %s, %s!" % (profile.get("favorite_color", "unknown"), name)
            elif "name" in last_msg.lower():
                resp_text = "Your name is %s!" % name
            else:
                resp_text = "Nice to see you again, %s! (offline stub)" % name
        else:
            resp_text = "Nice to meet you! (offline stub)"
        response = AIMessage(content=resp_text)

    # Extract and save any new facts from the last user message
    last_user = ""
    for m in reversed(state["messages"]):
        if hasattr(m, "type") and m.type == "human":
            last_user = m.content
            break

    if last_user:
        new_facts = extract_facts(last_user)
        if new_facts:
            merged = dict(profile)
            merged.update(new_facts)
            store.put(user_ns, "profile", merged)

    return {"messages": [response]}
```

```python
# --- Section 5: building the graph --------------------------------------------
graph_builder = StateGraph(ChatState)
graph_builder.add_node("chatbot", chatbot_with_memory)
graph_builder.add_edge(START, "chatbot")
graph_builder.add_edge("chatbot", END)

checkpointer = MemorySaver()
graph = graph_builder.compile(checkpointer=checkpointer, store=store)
print("Graph compiled with InMemoryStore backend.")
```

```python
# --- Section 6: Session 1 -- learning user preferences -------------------------
# Alice introduces herself and shares preferences.
# The chatbot saves these facts to the persistent store.
session1_config = {
    "configurable": {
        "thread_id": "alice-s1",
        "user_id": "alice",
    }
}

print("=== Session 1: Alice introduces herself ===\n")
turns_s1 = [
    "Hi, my name is Alice.",
    "My favorite color is purple.",
    "I enjoy hiking and painting.",
]
for turn in turns_s1:
    result = graph.invoke(
        {"messages": [("user", turn)]},
        session1_config,
    )
    print("user: %s" % turn)
    print("bot : %s" % result["messages"][-1].content)
    print()

# Verify what was stored
alice_profile = store.get(("user", "alice"), "profile")
print("Alice's stored profile:", json.dumps(alice_profile.value, indent=2))
```

```python
# --- Section 7: Session 2 -- remembering across threads -----------------------
# A brand-new thread. The chatbot should still remember Alice.
session2_config = {
    "configurable": {
        "thread_id": "alice-s2",    # different thread!
        "user_id": "alice",          # same user
    }
}

print("=== Session 2: New thread, same user ===\n")
result = graph.invoke(
    {"messages": [("user", "Hello again! Do you remember me?")]},
    session2_config,
)
print("user: Hello again! Do you remember me?")
print("bot : %s" % result["messages"][-1].content)
print()

# Ask a question that requires memory
result = graph.invoke(
    {"messages": [("user", "What is my favorite color?")]},
    session2_config,
)
print("user: What is my favorite color?")
print("bot : %s" % result["messages"][-1].content)
print()
```

```python
# --- Section 8: user isolation demo -------------------------------------------
# Bob should NEVER see Alice's preferences.
print("=== User Isolation Demo ===\n")
bob_config = {
    "configurable": {
        "thread_id": "bob-s1",
        "user_id": "bob",
    }
}

result = graph.invoke(
    {"messages": [("user", "Hi, my name is Bob. My favorite color is green.")]},
    bob_config,
)
print("user (Bob): Hi, my name is Bob. My favorite color is green.")
print("bot       : %s" % result["messages"][-1].content)
print()

# Check stored profiles
alice_profile = store.get(("user", "alice"), "profile")
bob_profile = store.get(("user", "bob"), "profile")
print("Alice profile:", alice_profile.value if alice_profile else "None")
print("Bob profile  :", bob_profile.value if bob_profile else "None")
print("\nlesson: each user has their own isolated namespace in the store.")
```

```python
# --- Section 9: updating stored facts -----------------------------------------
# When a user shares new information, we merge it with existing facts.
print("\n=== Updating Facts ===\n")
update_config = {
    "configurable": {
        "thread_id": "alice-s3",
        "user_id": "alice",
    }
}

result = graph.invoke(
    {"messages": [("user", "Actually, my favorite color is now teal.")]},
    update_config,
)
print("user: Actually, my favorite color is now teal.")
print("bot : %s" % result["messages"][-1].content)
print()

alice_profile = store.get(("user", "alice"), "profile")
print("Updated Alice profile:", json.dumps(alice_profile.value, indent=2))
print("lesson: the store merges new facts with existing ones.")
```

```python
# --- Section 10: inspect all stored data --------------------------------------
print("=== All Stored Data ===\n")
all_items = store.search(("user",))
print("Total user profiles: %d" % len(all_items))
for item in all_items:
    print("  user=%s: %s" % (item.key, json.dumps(item.value)))

print("\n" + "=" * 60)
print("Long-Term Memory Summary:")
print("  InMemoryStore: key-value store organized by namespace tuples")
print("  Namespaces enable user-level isolation (alice never sees bob's data)")
print("  Facts persist across threads within the same process")
print("  For production, use Redis or database-backed stores")
print("  Combine with MemorySaver for per-thread conversation state")
print("=" * 60)
```
